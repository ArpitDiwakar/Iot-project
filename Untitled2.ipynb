```python

import tensorflow as tf

print("TensorFlow version:", tf.__version__)
print("GPU available:", tf.config.list_physical_devices("GPU"))

if tf.config.list_physical_devices("GPU"):
    print("Great! Your notebook can use a GPU.")
else:
    print("Using CPU. You can enable a GPU in Runtime settings.")
```

```python

import os
import zipfile
import urllib.request

ZIP_PATH = "/content/cats_and_dogs_filtered.zip"
BASE_DIR = "/content"

# Alternative download URL
URL = (
    "https://storage.googleapis.com/"
    "tensorflow-1-public/course2/cats_and_dogs_filtered.zip"
)

print("Downloading dataset...")

urllib.request.urlretrieve(URL, ZIP_PATH)

print("Download completed!")
print("ZIP size:", round(os.path.getsize(ZIP_PATH) / (1024 * 1024), 2), "MB")

print("Extracting dataset...")

with zipfile.ZipFile(ZIP_PATH, "r") as zip_ref:
    zip_ref.extractall(BASE_DIR)

dataset_dir = "/content/cats_and_dogs_filtered"
train_dir = os.path.join(dataset_dir, "train")
val_dir = os.path.join(dataset_dir, "validation")

assert os.path.isdir(train_dir), "Training folder not found!"
assert os.path.isdir(val_dir), "Validation folder not found!"

print("\nDataset ready!")
print("Training path:", train_dir)
print("Validation path:", val_dir)

for split_dir in [train_dir, val_dir]:
    print(f"\n{os.path.basename(split_dir)}:")
    for cls in ["cats", "dogs"]:
        folder = os.path.join(split_dir, cls)
        print(cls, ":", len(os.listdir(folder)), "images")
```

```python

IMAGE_SIZE = (160, 160)
BATCH_SIZE = 32
SEED = 123

train_ds = tf.keras.utils.image_dataset_from_directory(
    str(train_dir),
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    class_names=["cats", "dogs"],
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    str(val_dir),
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    class_names=["cats", "dogs"],
    shuffle=False
)

# Save class names before applying prefetch
class_names = train_ds.class_names
print("Class order:", class_names)

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

print("Image preprocessing completed successfully!")
```

```python

from tensorflow import keras
from tensorflow.keras import layers

data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05)
])

model = keras.Sequential([
    keras.Input(shape=(160, 160, 3)),

    data_augmentation,
    layers.Rescaling(1.0 / 255),

    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, activation="relu"),
    layers.MaxPooling2D(),

    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.4),

    layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python

from pathlib import Path

OUTPUT_DIR = Path("/content/outputs")
OUTPUT_DIR.mkdir(exist_ok=True)

MODEL_PATH = OUTPUT_DIR / "cat_dog_cnn.keras"

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True
    ),
    keras.callbacks.ModelCheckpoint(
        filepath=str(MODEL_PATH),
        monitor="val_loss",
        save_best_only=True
    )
]

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=8,
    callbacks=callbacks
)

model.save(MODEL_PATH)

print("Training completed!")
print("Saved model:", MODEL_PATH)
```

```python

import matplotlib.pyplot as plt

loss, accuracy = model.evaluate(val_ds, verbose=1)

print(f"Validation loss: {loss:.4f}")
print(f"Validation accuracy: {accuracy * 100:.2f}%")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history.history["accuracy"], label="Training")
axes[0].plot(history.history["val_accuracy"], label="Validation")
axes[0].set_title("CNN Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()

axes[1].plot(history.history["loss"], label="Training")
axes[1].plot(history.history["val_loss"], label="Validation")
axes[1].set_title("CNN Loss")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()

plt.tight_layout()
plt.savefig("/content/outputs/training_history.png", dpi=150)
plt.show()
```

```python

from google.colab import files
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

uploaded = files.upload()

for filename in uploaded.keys():
    image = Image.open(filename).convert("RGB")
    resized = image.resize((160, 160))

    image_array = np.asarray(resized, dtype=np.float32)
    image_batch = np.expand_dims(image_array, axis=0)

    probability = float(model.predict(image_batch, verbose=0)[0][0])

    if probability >= 0.5:
        label = "Dog"
        confidence = probability
    else:
        label = "Cat"
        confidence = 1 - probability

    plt.figure(figsize=(5, 5))
    plt.imshow(image)
    plt.axis("off")
    plt.title(f"Prediction: {label} ({confidence * 100:.2f}%)")
    plt.show()

    print("Predicted class:", label)
    print(f"Model confidence: {confidence * 100:.2f}%")
```

```python

import gradio as gr
import numpy as np
from PIL import Image

def classify_image(image):
    if image is None:
        return {"Please upload an image": 1.0}

    image = image.convert("RGB").resize((160, 160))
    array = np.asarray(image, dtype=np.float32)
    batch = np.expand_dims(array, axis=0)

    probability = float(model.predict(batch, verbose=0)[0][0])

    return {
        "Cat": float(1 - probability),
        "Dog": float(probability)
    }

demo = gr.Interface(
    fn=classify_image,
    inputs=gr.Image(type="pil", label="Upload a cat or dog image"),
    outputs=gr.Label(num_top_classes=2, label="Prediction"),
    title="🐱 Cat vs Dog Classifier 🐶",
    description=(
        "A CNN-based image classification project built "
        "with TensorFlow and Keras."
    )
)

demo.launch(share=False)
```
